```python
def perkalian(angka1, angka2):
    print(angka1, "x", angka2, "=", end="  ")
    hasil = angka1 * angka2
    for i in range(angka1-1):
        print(angka2, "+", end="  ")
    print(angka2, "=", hasil)

perkalian(6, 5)
perkalian(7, 10)
```

```python
mulai = int(input("Masukkan batas bawah: "))
selesai = int(input("Masukkan batas atas: "))

def ganjil(mulai, selesai):
    if mulai < selesai:
        angka = mulai + 1 if mulai % 2 == 0 else mulai
        while angka < selesai:
            print(angka, end="," if angka + 2 < selesai else ".")
            angka += 2
    else:
        angka = mulai - 1 if mulai % 2 == 0 else mulai
        while angka > selesai:
            print(angka, end="," if angka - 2 > selesai else ".")
            angka -= 2

ganjil(mulai, selesai)
```

```python
def hitungIPS():
    print("program perhitungan IPS mahasiswa")
    jumlah_matkul = int(input("Berapa jumlah mata kuliah? "))
    total_bobot = 0
    kredit_permatkul = 3
    for i in range(1, jumlah_matkul + 1):
        while True:
            huruf = input(f'nilai MK ke{i}: ')
            if huruf in ['A', 'B', 'C', 'D']:
                break
            else:
                print("Nilai harus menggunakan huruf kapital")
        if huruf == 'A':
            poin = 4
        elif huruf == 'B':
            poin = 3
        elif huruf == 'C':
            poin = 2
        elif huruf == 'D':
            poin = 1
        total_bobot += poin * kredit_permatkul
    ips = total_bobot / (jumlah_matkul * kredit_permatkul)
    print(f'Nilai IPS anda semester ini {ips:.2f}')

hitungIPS()
```
