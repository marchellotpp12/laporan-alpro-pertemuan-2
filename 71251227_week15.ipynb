```python
def cek_prima(bilangan, pembagi=2):
    if bilangan < 2:
        return False
    if pembagi * pembagi > bilangan:
        return True
    if bilangan % pembagi == 0:
        return False
    return cek_prima(bilangan, pembagi + 1)

masukan = int(input("Masukkan bilangan: "))

if cek_prima(masukan):
    print("Bilangan Prima")
else:
    print("Bukan Bilangan Prima")
```

```python
def cek_palindrom(kalimat):
    huruf = kalimat.lower()

    if len(huruf) <= 1:
        return True

    if huruf[0] != huruf[-1]:
        return False

    return cek_palindrom(huruf[1:-1])

masukan = input("Masukkan kata: ")

if cek_palindrom(masukan):
    print("Palindrom")
else:
    print("Bukan Palindrom")
```

```python
def jumlah_ganjil(batas):
    if batas == 1:
        return 1

    return batas + jumlah_ganjil(batas - 2)

batas_akhir = int(input("Bilangan ganjil terakhir: "))

print("Jumlah =", jumlah_ganjil(batas_akhir))
```

```python
def jumlah_digit(bilangan):
    if bilangan < 10:
        return bilangan

    return bilangan % 10 + jumlah_digit(bilangan // 10)

masukan = int(input("Masukkan bilangan: "))

print("Jumlah digit =", jumlah_digit(masukan))
```

```python
def cari_kombinasi(jumlah, ambil):
    if ambil == 0 or ambil == jumlah:
        return 1

    return cari_kombinasi(jumlah-1, ambil-1) + cari_kombinasi(jumlah-1, ambil)

jumlah_total = int(input("Masukkan n: "))
jumlah_ambil = int(input("Masukkan r: "))

print("Hasil kombinasi =", cari_kombinasi(jumlah_total, jumlah_ambil))
```
