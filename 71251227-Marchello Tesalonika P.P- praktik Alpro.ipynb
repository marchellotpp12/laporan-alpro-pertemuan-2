```python
try:
  nomor_bulan = int(input("Masukkan angka bulan: "))
  jumlah_hari = { 1: 31, 2: 29, 3: 31, 4: 30, 5: 31, 6: 30, 7: 31, 8: 31, 9: 30, 10: 31, 11: 30, 12: 31}

  if 1 <= nomor_bulan <= 12:
    print("Jumlah hari adalah", jumlah_hari[nomor_bulan])
  else:
    print("Bulan tidak valid")
except:
  print("Anda tidak memasukkan angka")
```

```python
try:
  panjang_1 = int(input("Masukkan sisi 1: "))
  panjang_2 = int(input("Masukkan sisi 2: "))
  panjang_3 = int(input("Masukkan sisi 3: "))

  if panjang_1 == panjang_2 == panjang_3:
    print("3 sisi sama")
  elif panjang_1 == panjang_2 or panjang_1 == panjang_3 or panjang_2 == panjang_3:
    print("2 sisi sama")
  else:
    print("Tidak ada yang sama")
except:
  print("Anda tidak memasukkan angka")
```
