```python
def tiga_terbesar(angka):
    unik = list(set(angka))
    unik.sort(reverse=True)
    return unik[:3]

angka = [4, 10, 7, 10, 3, 5, 8, 9]
print(tiga_terbesar(angka))
```

```python
def angka_done():
    angka = []

    while True:
        masuk = input('Masukkan angka (atau ketik "done" untuk selesai):')
        if masuk.lower() == 'done':
            break
        try:
            nilai = float(masuk)
            angka.append(nilai)
        except ValueError:
            print('Input tidak valid, masukkan bilangan atau "done"')
    
    if angka:
        rata_rata = sum(angka) / len(angka)
        print(f'Rata-rata bilangan yang dimasukkan adalah: {rata_rata}')
    else:
        print('Tidak ada bilangan yang dimasukkan')

angka_done()
```

```python
def kata_unik(nama_file):
    with open(nama_file, 'r', encoding='utf-8') as f:
        teks = f.read().lower()
    kata = teks.split()
    unik = set(kata)
    print('kata unik dalam file')
    for i in sorted(unik):
        print(i)

nama = input('Masukkan (namafile.txt): ')
kata_unik(nama)
```
