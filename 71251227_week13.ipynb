```python
def all_same(items):
    for item in items:
        if item != items[0]:
            return False
    return True

numbers = (90, 90, 90, 90)
print(all_same(numbers))
```

```python
data = ('Marchello Tesalonika ', '71251227', 'Pati, Jateng')

print('NIM    :', data[1])
print('NAMA   :', data[0])
print(f'ALAMAT : {data[2]}\n')

nim_tuple = tuple(data[1])
print(f'NIM : {nim_tuple}\n')

kata_pertama = data[0].split()[0]
sisa_nama = kata_pertama[1:]
sisa_nama_tuple = tuple(sisa_nama)
print(f'NAMA : {sisa_nama_tuple}\n')

nama_terbalik = tuple(data[0].split()[::-1])
print('NAMA TERBALIK:', nama_terbalik)
```

```python
file_name = input('Enter the file name .txt : ')

try:
    with open(file_name, 'r') as f:
        hours = {}
        for line in f:
            if line.startswith('From'):
                time = line.split()[5].split(':')[0]
                hours[time] = hours.get(time, 0) + 1

        for key in sorted(hours):
            print(key, hours[key])
except FileNotFoundError:
    print('File not found')
```
