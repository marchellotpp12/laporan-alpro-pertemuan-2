```python
def cek_angka(a, b, c):
  if a != b and a != c and b != c:
    if a + b == c or a + c == b or b + c == a:
      return True
  return False
print(cek_angka(10, 5, 15))
print(cek_angka(5, 10, 20))
```

```python
def cek_digit_akhir():
    try:
        bilangan_ke1 = int(input("Masukkan angka pertama: "))
        bilangan_ke2 = int(input("Masukkan angka kedua: "))
        bilangan_ke3 = int(input("Masukkan angka ketiga: "))
        
        x = bilangan_ke1 % 10
        y = bilangan_ke2 % 10
        z = bilangan_ke3 % 10

        if x == y or x == z or y == z:
            return True
        else:
            return False
    except eror:
        print("Input yang diberikan harus angka.")
        return cek_digit_ahir()
        
print(cek_digit_akhir())
```

```python
konversi_celcius_fahrenheit = lambda C: (9/5) * C + 32
konversi_celcius_reamur = lambda C: 0.8 * C

print("Input C = 100 F =", konversi_celcius_fahrenheit(100))
print("Input C = 80 R =", konversi_celcius_reamur(80))
print("Input C = 0 F =", konversi_celcius_fahrenheit(0))
```
